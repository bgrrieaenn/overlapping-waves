```python
# The old way, using the files Aurelien sent

file_ids = np.array(['ST2_C0_02_STR2_tsen_with_ERA5.nc',
                     'ST2_C0_03_TTL3_tsen_with_ERA5.nc',
                     'ST2_C0_04_TTL1_tsen_with_ERA5.nc',
                     'ST2_C0_05_TTL2_tsen_with_ERA5.nc',
                     'ST2_C0_06_STR1_tsen_with_ERA5.nc',
                     'ST2_C0_07_STR2_tsen_with_ERA5.nc',
                     'ST2_C0_08_STR2_tsen_with_ERA5.nc',
                     'ST2_C1_01_TTL5_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_02_STR1_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_03_TTL4_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_04_STR2_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_05_TTL3_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_06_TTL5_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_07_TTL4_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_08_STR1_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_09_STR2_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_10_TTL3_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_11_TTL1_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_12_STR4_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_13_STR1_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_15_TTL4_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_16_TTL5_TSEN_v01_with_ERA5.nc',
                     'ST2_C1_17_TTL3_TSEN_v01_with_ERA5.nc'])
flights = pd.DataFrame(columns = ['flight_id','time','latitude','longitude','altitude',
                                  'u','v','pressure','T_ERA5','N2_ERA5','superpressure'])

n_files = file_ids.size
for i in np.arange(n_files):
    
    # Load the data into a pandas dataframe
    print('File', i+1,'of',n_files,'files')
    file_path = str(f'../strateole2_from_aurelien/{file_ids[i]}')
    dataset = nc.Dataset(file_path)
    flight = pd.DataFrame(columns = flights.columns)
    flight['u'] = np.array(dataset.variables['u'])
    flight['v'] = np.array(dataset.variables['v'])
    flight['altitude'] = np.array(dataset.variables['alt'])
    flight['pressure'] = np.array(dataset.variables['pressure'])
    flight['T_ERA5'] = np.array(dataset.variables['T_ERA5'])
    flight['N2_ERA5'] = np.array(dataset.variables['N2_ERA5'])
    flight['latitude'] = np.array(dataset.variables['lat'])
    flight['longitude'] = np.array(dataset.variables['lon'])
    flight['superpressure'] = np.array(dataset.variables['superpressure'])
    flight['flight_id'] = file_ids[i][4:14]
    
    # Get the time in the right format
    time = pd.to_timedelta(np.array(dataset.variables['time']), 's')
    flight['time'] = np.datetime64('2019-11-01') + time
    
    # Append to the combined dataframe
    flights = flights.append(flight, ignore_index=True)
    
# Load in the Flight C0_01_STR1 data that I've added ERA-5 reanalysis data to
flight = pd.read_feather('../era5_T_N2/strateole2_flight_c0_01/ST2_C0_01_STR1_tsen_with_ERA5_Brian.feather')
flights = flights.append(flight, ignore_index=True)

# Sort data for each flight by time and save
flights = flights.sort_values(by=['flight_id', 'time']).reset_index(drop=True)
flights.to_feather('Strateole2_with_ERA5.feather') 
flights
```
